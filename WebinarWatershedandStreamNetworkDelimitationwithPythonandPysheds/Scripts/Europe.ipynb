```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors
import geopandas as gpd
from pysheds.grid import Grid
import mplleaflet
# %matplotlib inline

# Create the grid
grid = Grid.from_raster('../Rst/hyd_eu_dem_30s.tif')
# Read the DEM
dem = grid.read_raster('../Rst/hyd_eu_dem_30s.tif')

# Visualizar el DEM
def plotFigure(data, label, cmap=None):
    plt.figure(figsize=(8, 6))
    plt.imshow(data, cmap=cmap)
    plt.colorbar(label=label)
    plt.grid()
```

```python
print(dem.shape, dem.min(), dem.max())
plt.figure(figsize=(8, 6))
plt.imshow(dem[:-1, :-1])
plt.grid()
```

```python
plotFigure(dem, 'Elevation (m)', cmap='terrain')
```
